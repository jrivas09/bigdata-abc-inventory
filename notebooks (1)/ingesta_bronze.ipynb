```python
# 1. Definir la ruta del dataset público que ya existe dentro de Databricks
file_path = "/databricks-datasets/online_retail/data-001/data.csv"

# 2. Leer el archivo CSV directamente a un DataFrame de PySpark
df_raw = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
)

# 3. Guardar directamente en la Capa Bronze como Tabla Delta
df_raw.write.format("delta").mode("overwrite").saveAsTable("bronze_inventory_raw")

# 4. Mostrar los primeros 10 registros para verificar la lectura
display(spark.table("bronze_inventory_raw").limit(10))
```
